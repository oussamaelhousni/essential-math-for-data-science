```python
def derivative_at_x(function,x,epsilon):
    return (function(x + epsilon) - function(x))/epsilon
```

The derivative is:

$$
f'(x) = 2x
$$

At \(x = 2\):

$$
f'(2) = 2(2) = 4
$$

Using a very small value of \(h\):

$$
f'(x)
=
\lim_{h \to 0}
\frac{f(x+h)-f(x)}{h}
$$

Substituting \(x=2\):

$$
f'(2)
=
\lim_{h \to 0}
\frac{(2+h)^2 - 2^2}{h}
=
4
$$

```python
def x_2(x):
    return x ** 2
print(derivative_at_x(x_2,2,0.0000000000001))
```
