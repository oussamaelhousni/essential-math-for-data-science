```python
import numpy as np

basis = np.array([
    [3,0],
    [0,2]
])

v = np.array([1,1])

new_v = basis.dot(v)

print(new_v)
```

```python
from numpy import array
# Declare i-hat and j-hat
i_hat = array([2, 0])
j_hat = array([0, 3])
# compose basis matrix using i-hat and j-hat
# also need to transpose rows into columns
basis = array([i_hat, j_hat]).transpose()
# declare vector v
v = array([1,1])
# create new vector
# by transforming v with dot product
new_v = basis.dot(v)
print(new_v) # [2, 3]
```

```python
# solve a linear equation
from numpy import array
from numpy.linalg import inv
# 4x + 2y + 4z = 44
# 5x + 3y + 7z = 56
# 9x + 3y + 6z = 72

A = array([
    [4,2,4],
    [5,3,7],
    [9,3,6]
])

B = array([44,56,72])

A_inv = inv(A)

solution = A_inv.dot(B)

print(solution)
```
