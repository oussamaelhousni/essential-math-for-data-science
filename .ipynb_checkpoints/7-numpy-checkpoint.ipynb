# Exercise 1: Student Grades Analysis with NumPy

## Context

You are working for a school.  
The administration wants to analyze students' grades before building a reporting system using Pandas.

You receive the following data:

Each row represents a student.

Columns represent:

1. Math grade
2. Physics grade
3. Computer Science grade

```python

grades = np.array([
    [14, 16, 15],
    [10, 12, 13],
    [18, 17, 19],
    [9, 11, 10],
    [15, 14, 16],
    [13, 15, 12]
])

print(grades)
```

````python
# exercise 1 Q1
import numpy as np

grades = np.array([
    [14, 16, 15],
    [10, 12, 13],
    [18, 17, 19],
    [9, 11, 10],
    [15, 14, 16],
    [13, 15, 12]
],dtype=np.float64)

print(grades.shape)
print(grades.ndim)
print(grades.size)
print(grades.dtype)
````

````python
#exercise 1 Q2
print("Math grades : {}".format(grades[:,0]))
print("Physic grades : {}".format(grades[:,1]))
print("Computer Science grades : {}".format(grades[:,2]))

print(grades.mean(axis=0))
print(grades.max(axis=0))
print(grades.min(axis=0))
````

````python
# exercise 1 Q3
averages = grades.mean(axis=1)
min_grade = np.min(averages)
max_grade = np.max(averages)
print(averages)
print(min_grade)
print(max_grade)
````

````python
# exercise 1 Q4
math_grades = grades[:,0]
math_more_1 = grades[grades[:,0]>=15]
print(math_more_1)
````

````python
# exerise 1 q5
final_grades = grades.mean(axis=1)
np.column_stack((grades,final_grades))
````
