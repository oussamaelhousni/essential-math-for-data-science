```python
def fx(x):
    return x**2 + 1
```

```python
def approximate_integral(f,a,b,n):
    width = (b-a)/n
    integral = 0
    for i in range(n):
        x_i = a +  width * i + width / 2
        integral += f(x_i) * width
    return integral
```

```python
approximate_integral(fx,0,1,100000)
```

```python
from sympy import integrate,symbols
x = symbols("x")
f = x **2 + 1
print(integrate(f,(x,0,1)))
```
