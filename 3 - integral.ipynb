```python
def fx(x):
    return x**2 + 1
```

```python
def approximate_integral(f,a,b,n):
    width = (b-a)/n
    integral = 0
    for i in range(n):
        x_i = a +  width * i + width / 2
        integral += f(x_i) * width
    return integral
```

```python
approximate_integral(fx,0,1,100000)
```

# Riemann Integral

The **Riemann integral** represents the **area under a curve** by dividing the interval into many small rectangles and summing their areas.

$$
\int_a^b f(x)\,dx
=
\lim_{n \to \infty}
\sum_{i=1}^{n}
f(x_i)\,\Delta x
$$

## Where:

- \(a\) = start of the interval  
- \(b\) = end of the interval  
- \(n\) = number of rectangles  
- \(\Delta x = \frac{b-a}{n}\) = width of each rectangle  
- \(x_i\) = a chosen point inside the \(i\)-th interval  
- \(f(x_i)\) = height of the \(i\)-th rectangle  

## Intuition

We approximate the area under the curve using rectangles:

- **Few rectangles** → rough approximation  
- **More rectangles** → better approximation  
- **Infinitely many rectangles** → exact area  

Imagine trying to measure the area under a curve:

1. Split the interval into many small pieces  
2. Draw a rectangle on each piece  
3. Add the areas of all rectangles  

The more rectangles we use, the more accurate the approximation becomes.

## Rectangle Area

The area of one rectangle is:

$$
\text{Area}
=
\text{height}
\times
\text{width}
=
f(x_i)\Delta x
$$

The total approximate area is:

$$
\sum_{i=1}^{n}
f(x_i)\Delta x
$$

Taking the limit as the number of rectangles becomes infinitely large gives the exact area:

$$
\int_a^b f(x)\,dx
=
\lim_{n \to \infty}
\sum_{i=1}^{n}
f(x_i)\Delta x
$$

## Example

For:

$$
f(x)=x^2
$$

The area under the curve from \(0\) to \(1\) is:

$$
\int_0^1 x^2\,dx
=
\frac{1}{3}
$$

Using a Riemann sum:

$$
\int_0^1 x^2\,dx
=
\lim_{n\to\infty}
\sum_{i=1}^{n}
\left(\frac{i}{n}\right)^2
\cdot
\frac{1}{n}
=
\frac{1}{3}
$$

### Key Idea

> A **Riemann integral** is simply the idea of finding an area by adding infinitely many tiny rectangles.

```python
from sympy import limit, Sum, oo, symbols
x, n , i = symbols("x n i")
f = x ** 2 + 1
# boundries
a = 0
b = 1
delta = (b - a) / n
x_i = a + i * delta

# sum of all rectangles
all_rects = Sum(delta * f.subs(x,x_i),(i,0,n)).doit()

# riemman limit
integral = limit(all_rects,n,oo)

print(integral)
```
