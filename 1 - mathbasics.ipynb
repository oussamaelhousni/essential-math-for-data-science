### Subtitute a variable
$$
f_x = \frac{2}{x}
$$

Then:

$$
f_x(2) = \frac{2}{2} = 1
$$

```python
from sympy import symbols
x = symbols("x",real=True)
f_x = x ** 2
f_x_1 = f_x.subs(x,2)
f_x_1
```

### Simplification

```python
from sympy import simplify
x = symbols("x")
expr = x **2 - 1
print(simplify(expr))
```

### Factorization

```python
from sympy import factor
x = symbols("x")
expr = x ** 2 + 2*x + 1
print(factor(expr))
```

### Expansion

```python
from sympy import expand
x = symbols("x")
f_x = (x - 1) ** 2
print(expand(f_x))
```

### Derivative

$$
f(x) = \log(x^2)
$$

Derivative:

$$
\frac{d}{dx}\log(x^2)
=
\frac{2}{x}
$$

```python
from sympy import diff,symbols,log,exp
x = symbols("x")
f = log(x **2)
print(diff(f,x))
```

### Partial derivative

$$
f(x, y) = 2x + 2y
$$

Partial derivative with respect to \(x\):

$$
\frac{\partial}{\partial x}(2x + 2y) = 2
$$

```python
from sympy import diff,symbols
x,y = symbols("x y")
f = 2 * x + 2 * y
print(diff(f,x))
```

### Integration

$$
\int_0^1 x^2 \, dx
=
\left[\frac{x^3}{3}\right]_0^1
=
\frac{1}{3}
$$

```python
from sympy import symbols, integrate
x = symbols("x")
f = x ** 2
result = integrate(f,(x,0,1))
print(result)
```

### Limits
$$
\lim_{n \to \infty} \left(1 + \frac{1}{n}\right)^n
= e \approx 2.71828169413\ldots
$$

```python
from sympy import limit, symbols,oo
n = symbols("n",integer=True)
u = (1 + (1/n)) ** n
print(limit(u,n,oo))
```
