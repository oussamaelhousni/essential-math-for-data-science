# Binomial Probability

The probability that an event with probability of success **p** gives exactly **k successes** out of **n trials** is:

$$
P(X = k) = {n \choose k} p^k (1-p)^{n-k}
$$

Where:

- **n** = total number of trials
- **k** = number of successes
- **p** = probability of success
- **1 - p** = probability of failure

```python
from scipy.stats import binom
n = 10
p = 0.9
for k in range(n+1):
    proba = binom.pmf(k,n,p)
    print("k = {1}, p[X=k] = {1}".format(k,proba))
```

# Beta Distribution

The **Beta Distribution** models the unknown probability of success **p**.

It answers:

> Given the observed data, what is the likely value of **p**?

## Formula

$$
f(p) =
\frac{p^{\alpha - 1}(1-p)^{\beta - 1}}
{B(\alpha,\beta)}
$$

Where:

- $p$ = probability of success
- $\alpha$ = successes + 1
- $\beta$ = failures + 1

## Example

If we observe:

- 8 successes  
- 2 failures

Then:

$$
\alpha = 8 + 1 = 9
$$

$$
\beta = 2 + 1 = 3
$$

Result:

$$
Beta(9,3)
$$

This means the true probability of success is likely **high**, around **0.75 to 0.8**.

```python
from scipy.stats import beta

a = 8
b = 2
p = beta.cdf(.9,a,b)

# there 77.48% chance that the probabilty of p is .9 or less
print(p)
```

```python
# there 22.51% chance that the probabilty of p is greater or equal .9
p2 = 1 - beta.cdf(.9,a,b)
print(p2)
```

# Beta Distribution

The **Beta Distribution** models the uncertainty of an unknown probability of success **p**.

It answers:

> What is the likely value of the true probability of success?

![Beta Distribution](beta-distribution.png)

## Formula

$$
f(p)=
\frac{p^{\alpha-1}(1-p)^{\beta-1}}
{B(\alpha,\beta)}
$$

Where:

- $p$ = probability of success  
- $\alpha$ = successes + 1  
- $\beta$ = failures + 1  

## Example

Suppose we observe:

- **8 successes**
- **2 failures**

Then:

$$
\alpha = 9
$$

$$
\beta = 3
$$

So:

$$
p \sim Beta(9,3)
$$

## Probability Between 0.8 and 0.9

To calculate the probability that the true probability is between **0.8** and **0.9**:

$$
P(0.8 \le p \le 0.9)
=
CDF(0.9) - CDF(0.8)
$$

This works because the **CDF** gives the probability to the **left** of a value.

So:

- $CDF(0.9)$ = probability that $p \le 0.9$
- $CDF(0.8)$ = probability that $p \le 0.8$

Subtracting them gives the probability **between 0.8 and 0.9**.

```python
from scipy.stats import beta
a = 8
b = 2
p_lt_80 = beta.cdf(.8,a,b)
p_gt_90 = 1 - beta.cdf(.9,a,b)

p =  1 - p_lt_80 - p_gt_90

print(p)
```
